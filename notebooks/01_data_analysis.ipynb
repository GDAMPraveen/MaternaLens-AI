```python
# ==========================================
# CORRELATION HEATMAP
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt

# Work from the complete dataset
corr_df = df.copy()

# Convert DataValue to numeric
corr_df["NumericValue"] = pd.to_numeric(
    corr_df["DataValue"],
    errors="coerce"
)

# Keep only numeric observations
corr_df = corr_df.dropna(subset=["NumericValue"])

# Create Area × Indicator matrix
indicator_matrix = corr_df.pivot_table(
    index="AreaName",
    columns="Indicator",
    values="NumericValue",
    aggfunc="mean"
)

# Keep indicators with at least 3 areas having values
indicator_matrix = indicator_matrix.loc[
    :,
    indicator_matrix.notna().sum() >= 3
]

print("Indicators available for correlation:", indicator_matrix.shape[1])
print("Areas available:", indicator_matrix.shape[0])

# Calculate correlation
correlation_matrix = indicator_matrix.corr()

# Select strongest relationships for a readable heatmap
if correlation_matrix.shape[0] > 15:
    variance = indicator_matrix.var().sort_values(ascending=False)
    selected = variance.head(15).index
    correlation_matrix = correlation_matrix.loc[selected, selected]

# Plot
plt.figure(figsize=(14, 11))

plt.imshow(
    correlation_matrix,
    aspect="auto",
    interpolation="nearest"
)

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=90,
    fontsize=7
)

plt.yticks(
    range(len(correlation_matrix.index)),
    correlation_matrix.index,
    fontsize=7
)

plt.title(
    "Correlation Between SDG Indicators",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# AUTOMATIC INDICATOR RELATIONSHIP ANALYZER
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

relationship_df = df.copy()

# Convert DataValue to numeric
relationship_df["NumericValue"] = pd.to_numeric(
    relationship_df["DataValue"],
    errors="coerce"
)

# Count observations for every indicator
indicator_stats = (
    relationship_df
    .dropna(subset=["NumericValue"])
    .groupby("Indicator")
    .agg(
        observations=("NumericValue", "count"),
        mean=("NumericValue", "mean"),
        std=("NumericValue", "std")
    )
    .sort_values("observations", ascending=False)
)

print("INDICATOR RELATIONSHIP ANALYSIS")
print("=" * 60)

display(indicator_stats.head(20))

# --------------------------------------------------
# Find indicators with enough repeated observations
# --------------------------------------------------

eligible = indicator_stats[
    indicator_stats["observations"] >= 3
]

print(f"\nIndicators with >= 3 observations: {len(eligible)}")

# --------------------------------------------------
# Create Indicator × Time matrix
# --------------------------------------------------

if len(eligible) >= 2:

    selected_indicators = eligible.head(10).index

    relation_matrix = (
        relationship_df[
            relationship_df["Indicator"].isin(selected_indicators)
        ]
        .pivot_table(
            index="TimePeriod",
            columns="Indicator",
            values="NumericValue",
            aggfunc="mean"
        )
    )

    # Keep only columns with enough paired observations
    relation_matrix = relation_matrix.dropna(
        axis=1,
        thresh=3
    )

    print("\nRelationship matrix:")
    display(relation_matrix)

    if relation_matrix.shape[1] >= 2:

        correlation = relation_matrix.corr()

        print("\nCorrelation matrix:")
        display(correlation)

        # ------------------------------------------
        # Correlation heatmap
        # ------------------------------------------

        plt.figure(figsize=(12, 9))

        plt.imshow(
            correlation,
            aspect="auto",
            interpolation="nearest"
        )

        plt.colorbar(label="Correlation")

        plt.xticks(
            range(len(correlation.columns)),
            correlation.columns,
            rotation=90,
            fontsize=8
        )

        plt.yticks(
            range(len(correlation.index)),
            correlation.index,
            fontsize=8
        )

        # Write correlation values inside cells
        for i in range(len(correlation.index)):
            for j in range(len(correlation.columns)):
                value = correlation.iloc[i, j]

                if not np.isnan(value):
                    plt.text(
                        j,
                        i,
                        f"{value:.2f}",
                        ha="center",
                        va="center",
                        fontsize=8
                    )

        plt.title(
            "Indicator Relationship Heatmap",
            fontsize=16,
            fontweight="bold"
        )

        plt.xlabel("Indicators")
        plt.ylabel("Indicators")

        plt.tight_layout()
        plt.show()

    else:
        print(
            "\nNot enough indicators with comparable time-series "
            "observations for a meaningful correlation analysis."
        )

else:
    print(
        "\nThe current dataset does not contain enough repeated "
        "observations to calculate a meaningful indicator correlation."
    )
```

```python
# ==========================================
# DISTRIBUTION & OUTLIER ANALYSIS
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt

# Prepare numeric dataset
distribution_df = df.copy()

distribution_df["NumericValue"] = pd.to_numeric(
    distribution_df["DataValue"],
    errors="coerce"
)

distribution_df = distribution_df.dropna(
    subset=["NumericValue"]
)

# ------------------------------------------------
# Select indicators with enough observations
# ------------------------------------------------

indicator_counts = (
    distribution_df["Indicator"]
    .value_counts()
)

eligible_indicators = indicator_counts[
    indicator_counts >= 5
].head(8).index

box_data = []

for indicator in eligible_indicators:

    values = distribution_df.loc[
        distribution_df["Indicator"] == indicator,
        "NumericValue"
    ]

    box_data.append(values)

print("Indicators selected for distribution analysis:")
for indicator in eligible_indicators:
    print(
        f"• {indicator} "
        f"({indicator_counts[indicator]} observations)"
    )

# ------------------------------------------------
# Box Plot
# ------------------------------------------------

plt.figure(figsize=(14, 8))

plt.boxplot(
    box_data,
    labels=[
        str(indicator)[:35]
        for indicator in eligible_indicators
    ],
    patch_artist=True
)

plt.title(
    "SDG Indicator Distribution & Outlier Analysis",
    fontsize=16,
    fontweight="bold"
)

plt.ylabel("Numeric Value")

plt.xticks(
    rotation=45,
    ha="right",
    fontsize=9
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# OUTLIER DETECTION — IQR METHOD
# ==========================================

outlier_summary = []

for indicator in eligible_indicators:

    values = distribution_df.loc[
        distribution_df["Indicator"] == indicator,
        "NumericValue"
    ]

    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = values[
        (values < lower_bound) |
        (values > upper_bound)
    ]

    outlier_summary.append({
        "Indicator": indicator,
        "Observations": len(values),
        "Q1": round(Q1, 2),
        "Median": round(values.median(), 2),
        "Q3": round(Q3, 2),
        "Lower Bound": round(lower_bound, 2),
        "Upper Bound": round(upper_bound, 2),
        "Outliers": len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)

display(outlier_df)
```

```python
# ==========================================
# MATERNALENS AI — INSIGHT ENGINE
# ==========================================

import pandas as pd
import numpy as np

# ------------------------------------------
# 1. Prepare SDG 3.1 data
# ------------------------------------------

insight_df = df[
    df["Target"].str.startswith("3.1:", na=False)
].copy()

insight_df["Value"] = pd.to_numeric(
    insight_df["DataValue"],
    errors="coerce"
)

# ------------------------------------------
# 2. Extract MMR
# ------------------------------------------

mmr_data = insight_df[
    insight_df["Indicator"].str.startswith(
        "3.1.1",
        na=False
    )
].sort_values("TimePeriod")

first_mmr = mmr_data.iloc[0]["Value"]
latest_mmr = mmr_data.iloc[-1]["Value"]

mmr_reduction = first_mmr - latest_mmr

mmr_reduction_pct = (
    mmr_reduction / first_mmr
) * 100

# ------------------------------------------
# 3. SDG target analysis
# ------------------------------------------

sdg_target = 70

target_gap = latest_mmr - sdg_target

required_reduction_pct = (
    target_gap / latest_mmr
) * 100

# ------------------------------------------
# 4. Extract maternal-health indicators
# ------------------------------------------

def get_indicator_value(prefix):
    rows = insight_df[
        insight_df["Indicator"].str.startswith(
            prefix,
            na=False
        )
    ]

    if len(rows) == 0:
        return np.nan

    return rows.iloc[0]["Value"]


skilled_5yr = get_indicator_value("3.1.2")
skilled_1yr = get_indicator_value("3.1.3")
antenatal = get_indicator_value("3.1.4")

# ------------------------------------------
# 5. Coverage gap
# ------------------------------------------

if not np.isnan(skilled_1yr) and not np.isnan(antenatal):
    coverage_gap = skilled_1yr - antenatal
else:
    coverage_gap = np.nan

# ------------------------------------------
# 6. Generate structured insights
# ------------------------------------------

insights = []

insights.append({
    "Category": "MMR Trend",
    "Metric": "Initial MMR",
    "Value": first_mmr,
    "Unit": "per 100,000 live births",
    "Interpretation":
        f"Initial recorded MMR was {first_mmr:.0f}."
})

insights.append({
    "Category": "MMR Trend",
    "Metric": "Latest MMR",
    "Value": latest_mmr,
    "Unit": "per 100,000 live births",
    "Interpretation":
        f"Latest recorded MMR was {latest_mmr:.0f}."
})

insights.append({
    "Category": "Progress",
    "Metric": "MMR Reduction",
    "Value": mmr_reduction,
    "Unit": "per 100,000 live births",
    "Interpretation":
        f"MMR decreased by {mmr_reduction:.0f} "
        f"from the earliest to latest observation."
})

insights.append({
    "Category": "Progress",
    "Metric": "MMR Reduction Percentage",
    "Value": mmr_reduction_pct,
    "Unit": "%",
    "Interpretation":
        f"MMR declined by {mmr_reduction_pct:.2f}% "
        f"over the observed period."
})

insights.append({
    "Category": "SDG Target",
    "Metric": "Distance From Target",
    "Value": target_gap,
    "Unit": "per 100,000 live births",
    "Interpretation":
        f"Latest MMR is {target_gap:.0f} above "
        f"the SDG 3.1 target of less than 70."
})

insights.append({
    "Category": "SDG Target",
    "Metric": "Additional Reduction Required",
    "Value": required_reduction_pct,
    "Unit": "%",
    "Interpretation":
        f"A further reduction of approximately "
        f"{required_reduction_pct:.2f}% from the "
        f"latest value would be needed to reach 70."
})

if not np.isnan(skilled_1yr):

    insights.append({
        "Category": "Healthcare Coverage",
        "Metric": "Skilled Birth Attendance",
        "Value": skilled_1yr,
        "Unit": "%",
        "Interpretation":
            f"Skilled birth attendance coverage "
            f"was {skilled_1yr:.1f}%."
    })

if not np.isnan(antenatal):

    insights.append({
        "Category": "Healthcare Coverage",
        "Metric": "Antenatal Care 4+ Visits",
        "Value": antenatal,
        "Unit": "%",
        "Interpretation":
            f"Antenatal care coverage for four or "
            f"more visits was {antenatal:.1f}%."
    })

if not np.isnan(coverage_gap):

    insights.append({
        "Category": "Healthcare Gap",
        "Metric": "Attendance vs Antenatal Gap",
        "Value": coverage_gap,
        "Unit": "percentage points",
        "Interpretation":
            f"The skilled attendance measure exceeds "
            f"antenatal care coverage by "
            f"{coverage_gap:.1f} percentage points."
    })

# ------------------------------------------
# 7. Convert to DataFrame
# ------------------------------------------

insight_table = pd.DataFrame(insights)

print("🧠 MATERNA LENS — STRUCTURED INSIGHTS")
print("=" * 70)

display(insight_table)
```

```python
# ==========================================
# MATERNALENS AI — LOCAL DATA Q&A
# ==========================================

def maternalens_qa(question):

    question = question.lower().strip()

    # ------------------------------------------
    # MMR / maternal mortality
    # ------------------------------------------

    if "latest mmr" in question or "current mmr" in question:
        return (
            f"The latest recorded MMR in the dataset is "
            f"{latest_mmr:.0f} per 100,000 live births."
        )

    if "initial mmr" in question or "starting mmr" in question:
        return (
            f"The earliest recorded MMR in the dataset is "
            f"{first_mmr:.0f} per 100,000 live births."
        )

    if "reduction" in question and "mmr" in question:
        return (
            f"MMR decreased from {first_mmr:.0f} to "
            f"{latest_mmr:.0f}, a reduction of "
            f"{mmr_reduction:.0f} per 100,000 live births "
            f"({mmr_reduction_pct:.2f}%)."
        )

    # ------------------------------------------
    # SDG target
    # ------------------------------------------

    if "target" in question or "sdg 3.1" in question:

        return (
            f"The SDG 3.1 target is an MMR below "
            f"70 per 100,000 live births. The latest "
            f"recorded value is {latest_mmr:.0f}, "
            f"which is {target_gap:.0f} above the target."
        )

    # ------------------------------------------
    # Antenatal care
    # ------------------------------------------

    if "antenatal" in question or "anc" in question:

        return (
            f"Antenatal care coverage for four or more "
            f"visits is {antenatal:.1f}%."
        )

    # ------------------------------------------
    # Skilled birth attendance
    # ------------------------------------------

    if "skilled" in question or "birth attendance" in question:

        return (
            f"Recorded skilled birth attendance coverage "
            f"is {skilled_birth:.1f}%."
        )

    # ------------------------------------------
    # Coverage gap
    # ------------------------------------------

    if "gap" in question:

        return (
            f"The gap between skilled birth attendance "
            f"and antenatal care coverage is "
            f"{coverage_gap:.1f} percentage points."
        )

    # ------------------------------------------
    # General summary
    # ------------------------------------------

    if (
        "summary" in question
        or "overview" in question
        or "overall" in question
    ):

        return (
            f"The dataset shows improvement in India's "
            f"recorded maternal mortality ratio, falling "
            f"from {first_mmr:.0f} to {latest_mmr:.0f} "
            f"per 100,000 live births. However, the latest "
            f"value remains {target_gap:.0f} above the "
            f"SDG 3.1 target of less than 70. Skilled "
            f"birth attendance is {skilled_birth:.1f}%, "
            f"while antenatal care coverage for four or "
            f"more visits is {antenatal:.1f}%."
        )

    return (
        "I could not find a matching analysis for that "
        "question. Try asking about MMR, the SDG 3.1 "
        "target, antenatal care, skilled birth attendance, "
        "the reduction in MMR, or the coverage gap."
    )


# ==========================================
# TEST QUESTIONS
# ==========================================

test_questions = [
    "What is the latest MMR?",
    "How much has MMR reduced?",
    "How far are we from the SDG 3.1 target?",
    "What is the antenatal care coverage?",
    "What is the gap between skilled attendance and antenatal care?",
    "Give me an overall summary"
]

for question in test_questions:

    print(f"\nQ: {question}")
    print(f"AI: {maternalens_qa(question)}")
```
